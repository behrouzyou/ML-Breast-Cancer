# Helper Methods

## A standard evaluation function (for all models)

```python
def evaluate_model(name, model, X_tr, X_te, y_tr, y_te):
    model.fit(X_tr, y_tr)
    y_pred = model.predict(X_te)

    # Some models have predict_proba, some have decision_function
    y_score = None
    if hasattr(model, "predict_proba"):
        y_score = model.predict_proba(X_te)[:, 1]
    elif hasattr(model, "decision_function"):
        y_score = model.decision_function(X_te)

    results = {
        "model": name,
        "accuracy": accuracy_score(y_te, y_pred),
        "precision": precision_score(y_te, y_pred, zero_division=0),
        "recall": recall_score(y_te, y_pred, zero_division=0),
        "f1": f1_score(y_te, y_pred, zero_division=0),
        "roc_auc": roc_auc_score(y_te, y_score) if y_score is not None else np.nan
    }

    cm = confusion_matrix(y_te, y_pred)
    return results, cm
```

## Metrics Table

```python
def show_metrics_table(metrics_dict):
    cleaned = {}

    for k, v in metrics_dict.items():
        if isinstance(v, np.generic):
            cleaned[k] = float(v)
        else:
            cleaned[k] = v

    df = pd.DataFrame(cleaned.items(), columns=["Metric", "Value"])
    return df
```

## Confusion Matrix

```python
def show_confusion_matrix_table(cm):
    tn, fp, fn, tp = cm.ravel()

    df = pd.DataFrame(
        [
            [tn, fp],
            [fn, tp]
        ],
        columns=["Predicted Negative", "Predicted Positive"],
        index=["Actual Negative", "Actual Positive"]
    )

    print("\nDetails:")
    print(f"TN = {tn}")
    print(f"FP = {fp}")
    print(f"FN = {fn}")
    print(f"TP = {tp}")
    return df
```

## Confusion Matrix Analysis

Confusion Matrix display function + FNR/FPR calcuation

```python
def analyze_confusion_matrix(name, cm):
    tn, fp, fn, tp = cm.ravel()

    analysis = {
        "model": name,
        "TP": tp,
        "TN": tn,
        "FP": fp,
        "FN": fn,
        "FN_rate": fn / (fn + tp),
        "FP_rate": fp / (fp + tn)
    }

    return analysis
```

# Threshold Tuning

## Getting the LightGBM Output Probability

```python
lgbm.fit(X_train, y_train)

y_proba = lgbm.predict_proba(X_test)[:,1]
```

## Evaluation Function with threshold

```python
def evaluate_with_threshold(y_true, y_proba, threshold):
    y_pred = (y_proba >= threshold).astype(int)
    cm = confusion_matrix(y_true, y_pred)

    tn, fp, fn, tp = cm.ravel()

    return {
        "threshold": threshold,
        "TP": tp,
        "TN": tn,
        "FP": fp,
        "FN": fn,
        "recall": tp / (tp + fn),
        "precision": tp / (tp + fp) if (tp + fp) > 0 else 0,
        "fn_rate": fn / (fn + tp)
    }
```

## Testing Different thresholds

```python
thresholds = np.arange(0.1, 0.91, 0.05)

threshold_results = pd.DataFrame([
    evaluate_with_threshold(y_test,y_proba,t)
    for t in thresholds
])

threshold_results
```

### Choosing the optimal Threshold

```python
threshold_results.sort_values(by="FN")
```

## Final evaluation with new Threshold

```python
best_threshold = 0.10

y_predict_final = (y_proba >= best_threshold).astype(int)
cm_final = confusion_matrix(y_test,y_predict_final)

show_confusion_matrix_table(cm_final)
```

# Cross Validaiton with fixed threshold

```python
from sklearn.model_selection import StratifiedKFold

skf = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)
```

```python
THRESHOLD = 0.10

cv_results = []

for fold, (train_idx, val_idx) in enumerate(skf.split(X, y), start=1):
    X_tr, X_val = X.iloc[train_idx], X.iloc[val_idx]
    y_tr, y_val = y.iloc[train_idx], y.iloc[val_idx]

    model = LGBMClassifier(
        n_estimators=300,
        learning_rate=0.05,
        max_depth=-1,
        random_state=42,
        verbose=-1
    )

    model.fit(X_tr, y_tr)

    y_proba_val = model.predict_proba(X_val)[:, 1]
    y_pred_val = (y_proba_val >= THRESHOLD).astype(int)

    tn, fp, fn, tp = confusion_matrix(y_val, y_pred_val).ravel()

    cv_results.append({
        "fold": fold,
        "TP": tp,
        "TN": tn,
        "FP": fp,
        "FN": fn,
        "recall": tp / (tp + fn),
        "precision": tp / (tp + fp) if (tp + fp) > 0 else 0,
        "fn_rate": fn / (fn + tp)
    })

cv_df = pd.DataFrame(cv_results)
cv_df
```

# Inference Function

```python
def medical_inference(model, X_input, threshold=0.10):
    """
    Medical-safe inference function.
    
    Returns:
    - prediction (0=Benign, 1=Malignant)
    - probability
    - risk_label
    """
    proba = model.predict_proba(X_input)[:, 1]
    prediction = (proba >= threshold).astype(int)

    risk_label = np.where(
        proba >= threshold,
        "High Risk (Review Required)",
        "Low Risk"
    )

    return pd.DataFrame({
        "probability_malignant": proba,
        "prediction": prediction,
        "risk_label": risk_label
    })
```
